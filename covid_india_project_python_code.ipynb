# 7.Calculating  outliers using IQR

```python
outlier_columns = [
    "new_cases",
    "new_recovered",
    "new_deaths",
    "active_cases",
    "totalsamples"
]
```

```python
outlier_summary = []

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    outlier_summary.append({
        "Column": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower_Bound": lower_bound,
        "Upper_Bound": upper_bound,
        "Outlier_Count": len(outliers),
        "Outlier_Percentage": round(
            len(outliers) / df[col].notna().sum() * 100, 2
        )
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

# 8.Exploratory Data Analysis

##  Creating daily India-level data

```python
daily_india = (
    df.groupby("date", as_index=False)
      .agg({
          "confirmed": "sum",
          "new_cases": "sum",
          "active_cases": "sum",
          "cured": "sum",
          "new_recovered": "sum",
          "deaths": "sum",
          "new_deaths": "sum",
          "totalsamples": "sum",
          "positive": "sum",
          "total_doses": "sum",
          "first_dose": "sum",
          "second_dose": "sum",
          "total_individuals_vaccinated": "sum"
      })
)

daily_india.head()
```

```python
# ============================================
# COMPARE FIRST AND SECOND WAVE
# ============================================

wave1 = daily_india[
    (daily_india["date"] >= "2020-06-01") &
    (daily_india["date"] <= "2020-12-31")
]

wave2 = daily_india[
    (daily_india["date"] >= "2021-03-01") &
    (daily_india["date"] <= "2021-07-31")
]

wave_comparison = pd.DataFrame({
    "Metric": [
        "Peak New Cases",
        "Peak New Deaths",
        "Peak Active Cases"
    ],

    "Wave 1": [
        wave1["new_cases"].max(),
        wave1["new_deaths"].max(),
        wave1["active_cases"].max()
    ],

    "Wave 2": [
        wave2["new_cases"].max(),
        wave2["new_deaths"].max(),
        wave2["active_cases"].max()
    ]
})

wave_comparison
```

## EDA 1:COVID Trend & Wave Analysis

###  *India experienced two major COVID-19 waves during the study period.

### DAILY CASES WITH PEAK IDENTIFICATION

```python
# ============================================
# DAILY CASES WITH PEAK IDENTIFICATION
# ============================================

peak_case_row = daily_india.loc[
    daily_india["new_cases"].idxmax()
]

plt.figure(figsize=(14, 6))

plt.plot(
    daily_india["date"],
    daily_india["new_cases"]
)

plt.scatter(
    peak_case_row["date"],
    peak_case_row["new_cases"],
    s=80
)

plt.annotate(
    f"Peak: {peak_case_row['new_cases']:,}\n"
    f"{peak_case_row['date'].strftime('%d %b %Y')}",
    xy=(
        peak_case_row["date"],
        peak_case_row["new_cases"]
    ),
    xytext=(30, -60),
    textcoords="offset points",
    arrowprops=dict(arrowstyle="->")
)

plt.title("Daily New COVID-19 Cases in India")
plt.xlabel("Date")
plt.ylabel("New Cases")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### *The peak Daily new cases in India is about 414,188 which was recorded on 07-may-2021

### Peak New cases:wave comparision

```python
# ============================================
# EDA 1 — PEAK NEW CASES: WAVE COMPARISON
# ============================================

wave_labels = ["Wave 1", "Wave 2"]
case_values = [97894, 414188]

plt.figure(figsize=(8, 5))

bars = plt.bar(wave_labels, case_values)

plt.title("Peak Daily New COVID-19 Cases: Wave 1 vs Wave 2")
plt.xlabel("COVID-19 Wave")
plt.ylabel("Peak New Cases")

# Display values on bars
for bar, value in zip(bars, case_values):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:,}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

### *The first wave peaked at 97,894 daily cases, while the second wave reached 414,188, representing a 323.1% increase.

### PEAK NEW DEATHS: WAVE COMPARISON

```python
# ============================================
# EDA 1 — PEAK NEW DEATHS: WAVE COMPARISON
# ============================================

death_values = [2003, 6148]

plt.figure(figsize=(8, 5))

bars = plt.bar(wave_labels, death_values)

plt.title("Peak Daily COVID-19 Deaths: Wave 1 vs Wave 2")
plt.xlabel("COVID-19 Wave")
plt.ylabel("Peak New Deaths")

# Display values on bars
for bar, value in zip(bars, death_values):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{value:,}",
        ha="center",
        va="bottom"
    )

plt.tight_layout()
plt.show()
```

### *Peak daily deaths increased from 2,003 to 6,148, a 206.8% increase.

### Normalized wave comparision

```python
# ============================================
# NORMALIZED WAVE COMPARISON
# ============================================

metrics = [
    "Peak New Cases",
    "Peak New Deaths",
    "Peak Active Cases"
]

wave1 = [97894, 2003, 1017754]
wave2 = [414188, 6148, 3745237]

# Convert Wave 1 to 100
wave1_index = [100, 100, 100]

wave2_index = [
    (wave2[0] / wave1[0]) * 100,
    (wave2[1] / wave1[1]) * 100,
    (wave2[2] / wave1[2]) * 100
]

x = np.arange(len(metrics))
width = 0.35

plt.figure(figsize=(10, 6))

bars1 = plt.bar(
    x - width/2,
    wave1_index,
    width,
    label="Wave 1"
)

bars2 = plt.bar(
    x + width/2,
    wave2_index,
    width,
    label="Wave 2"
)

plt.title("Normalized Relative Severity of COVID-19 Waves")
plt.xlabel("Metric")
plt.ylabel("Relative Peak Index (Wave 1 = 100)")
plt.xticks(x, metrics)
plt.legend()

# Add values
for bars in [bars1, bars2]:
    for bar in bars:
        height = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width()/2,
            height,
            f"{height:.0f}",
            ha="center",
            va="bottom"
        )

plt.tight_layout()
plt.show()
```

### *The second wave therefore created substantially greater case and active-case burden than the first wave.
